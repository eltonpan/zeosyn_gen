```python
from rdkit import Chem
from rdkit.Chem import AllChem, Draw
```

```python
# Define SMILES and labels
smiles_list = [
    "C1C[N+]2(CCN1CC2)CCCC[N+]3(CCN4CC3)CC4",
    "C1C[N+]2(CCN1CC2)CCCCC[N+]3(CCN4CC3)CC4",
    "C1C[N+]2(CCC1CC2)CCCC[N+]3(CCC4CC3)CC4",
    "C1C[N+]2(CCC1CC2)CCCCC[N+]3(CCC4CC3)CC4",
]

labels = [
    "DiDABCO-C4",
    "DiDABCO-C5",
    "Diquinuclidinium-C4",
    "Diquinuclidinium-C5",
]

cleaned_smiles = []
for smiles in smiles_list:
  mol = Chem.MolFromSmiles(smiles)
  smiles = Chem.MolToSmiles(mol)
  cleaned_smiles.append(smiles)

# Generate RDKit mol objects with 2D coordinates
mols = []
for smi in cleaned_smiles:
    mol = Chem.MolFromSmiles(smi)
    Chem.Kekulize(mol, clearAromaticFlags=True)
    AllChem.Compute2DCoords(mol)
    mols.append(mol)

# Display grid inline
img = Draw.MolsToGridImage(mols, molsPerRow=2, subImgSize=(350, 300), legends=labels)
img
```
